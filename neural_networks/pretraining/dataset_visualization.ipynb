```python
def visualize_feature_maps( vision_model,  image_tensor : torch.Tensor):
    """
    Visualizes the output of each layer in the CNN encoder.
    image_tensor: [1, C, H, W]
    """
    vision_model.eval()
    x = image_tensor.clone().detach()

    for i, layer in enumerate(vision_model.encoder):
        x = layer(x)  # forward through layer
        if isinstance(layer, nn.ReLU) or isinstance(layer, nn.MaxPool2d):
            continue  # skip non-conv layers for clarity

        # Plot feature maps
        feature_map = x.squeeze(0)  # remove batch dim -> [C, H, W]
        num_channels = feature_map.shape[0]
        num_cols = 8
        num_rows = (num_channels + num_cols - 1) // num_cols

        fig, axes = plt.subplots(num_rows, num_cols, figsize=(num_cols * 1.5, num_rows * 1.5))
        for i in range(num_channels):
            ax = axes[i // num_cols, i % num_cols]
            ax.imshow(feature_map[i].detach().numpy(), cmap="viridis")
            ax.axis("off")
        plt.suptitle(f"Feature maps after layer {layer.__class__.__name__}")
        plt.tight_layout()
        plt.show()

visualize_feature_maps(model.vision_model, img)
```
